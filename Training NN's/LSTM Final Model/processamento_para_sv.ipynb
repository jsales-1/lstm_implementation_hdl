```python
def float_to_q16_16(value):
    """Converte float para Q16.16 (formato signed 32 bits)"""
    scaled = value * 65536
    rounded = int(round(scaled))
    
    # Satura para 32 bits signed
    if rounded > 2**31 - 1:
        rounded = 2**31 - 1
    elif rounded < -2**31:
        rounded = -2**31
    
    return rounded


def gerar_x_array(valores):
    """
    Gera código para inicializar array x a partir de valores float.
    
    Args:
        valores: Lista de floats
    
    Returns:
        String com as atribuições formatadas
    """
    q16_values = [float_to_q16_16(v) for v in valores]
    
    linhas = []
    linha_atual = ""
    
    for i, qval in enumerate(q16_values):
        if qval >= 0:
            atrib = f"x[{i}] =  32'sd{qval};"
        else:
            atrib = f"x[{i}] = -32'sd{abs(qval)};"
        
        linha_atual += f"{atrib}  "
        
        if (i + 1) % 8 == 0 or i == len(valores) - 1:
            linhas.append(linha_atual.strip())
            linha_atual = ""
    
    return "\n".join(linhas)


```

```python
def float_to_q16_16(value):
    """Converte float para Q16.16 (formato signed 32 bits)"""
    scaled = value * 65536
    rounded = int(round(scaled))

    if rounded > 2**31 - 1:
        rounded = 2**31 - 1
    elif rounded < -2**31:
        rounded = -2**31

    return rounded


def gerar_x_array_2d(valores):
    """
    Gera código:
    x[i][j] = ±32'sdN;

    Aceita:
        shape (N,M)
        ou (1,N,M)
    """

    # remove dimensão batch se existir
    if len(valores) == 1 and isinstance(valores[0][0], list):
        valores = valores[0]

    linhas = []

    for i, linha in enumerate(valores):

        linha_atual = ""

        for j, valor in enumerate(linha):

            qval = float_to_q16_16(valor)

            if qval >= 0:
                atrib = f"x[{i}][{j}] =  32'sd{qval};"
            else:
                atrib = f"x[{i}][{j}] = -32'sd{abs(qval)};"

            linha_atual += f"{atrib}   "

        linhas.append(linha_atual.rstrip())

    return "\n".join(linhas)
```
